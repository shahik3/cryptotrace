```python
path = "/Volumes/workspace/bronze/raw/"
```

```python
col_names = ["txId", "time_step"] + [f"local_{i:02d}" for i in range(1, 94)] + [f"agg_{i:02d}" for i in range(1, 73)]

features = (
    spark.read
    .option("header", "false")
    .csv(path + "elliptic_txs_features.csv")
    .toDF(*col_names)
)
```

```python
classes = spark.read.csv(path + "elliptic_txs_classes.csv", header=True)
edges = spark.read.csv(path + "elliptic_txs_edgelist.csv", header=True)

display(classes)
display(edges)
```

```python
from pyspark.sql.functions import current_timestamp, lit

features = features.withColumn("_ingested_at", current_timestamp())
features = features.withColumn("_source_file", lit("elliptic_txs_features.csv"))

classes = classes.withColumn("_ingested_at", current_timestamp())
classes = classes.withColumn("_source_file", lit("elliptic_txs_classes.csv"))

edges = edges.withColumn("_ingested_at", current_timestamp())
edges = edges.withColumn("_source_file", lit("elliptic_txs_edgelist.csv"))
```

```python
features.write.mode("overwrite").saveAsTable("bronze.elliptic_txs_features")
classes.write.mode("overwrite").saveAsTable("bronze.elliptic_txs_classes")
edges.write.mode("overwrite").saveAsTable("bronze.elliptic_txs_edgelist")
```

```python
print(spark.table("bronze.elliptic_txs_features").count())
print(spark.table("bronze.elliptic_txs_classes").count())
print(spark.table("bronze.elliptic_txs_edgelist").count())
```
