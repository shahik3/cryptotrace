## Step 1: Label counts by time step
We count how many transactions have each label at each time step.
Power BI will use this for charts.

```python
%sql
CREATE OR REPLACE TABLE gold.label_counts_by_time_step AS
SELECT time_step, label_name, COUNT(*) AS transactions
FROM silver.transactions
GROUP BY time_step, label_name
ORDER BY time_step, label_name
```

```python
%sql
SELECT SUM(transactions) AS total_transactions
FROM gold.label_counts_by_time_step
```

## Step 2: Headline numbers
One row with the main totals for the dashboard cards.

```python
%sql
CREATE OR REPLACE TABLE gold.summary_kpis AS
SELECT
  COUNT(*) AS total_transactions,
  COUNT(CASE WHEN label_name = 'illicit' THEN 1 END) AS illicit,
  COUNT(CASE WHEN label_name = 'licit' THEN 1 END) AS licit,
  COUNT(CASE WHEN label_name = 'unknown' THEN 1 END) AS unknown,
  ROUND(100.0 * COUNT(CASE WHEN label_name = 'illicit' THEN 1 END) / COUNT(*), 2) AS percent_illicit
FROM silver.transactions
```

```python
%sql
SELECT * FROM gold.summary_kpis
```
